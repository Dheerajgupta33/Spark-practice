```python
data1 = [
    (1, "Anil", "IT"),
    (2, "Vikas", "HR"),
    (3, "Nisha", "IT")
]

data2 = [
    (3, "Nisha", "IT"),
    (4, "Priya", "Finance"),
    (5, "Mohit", "HR")
]

df1 = spark.createDataFrame(data1, ["id", "name", "department"])
df2 = spark.createDataFrame(data2, ["id", "name", "department"])
```

```python
df1.createOrReplaceTempView("df1")
df2.createOrReplaceTempView("df2")       ## used for convert data frames to spark sql table
```

```python
spark.sql("SHOW TABLES").show()
```

```python
result = spark.sql("""
    SELECT * FROM df1
    UNION
    SELECT * FROM df2
""")

result.show()       ## in sql union removes dublicates 
```

```python
result = spark.sql("""
    SELECT * FROM df1
    UNION ALL
    SELECT * FROM df2
""")
                     
result.show()         ## in sql union all contains dublicates
```
