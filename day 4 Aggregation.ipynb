```python
import os
import sys

os.environ["PYSPARK_PYTHON"] = sys.executable
os.environ["PYSPARK_DRIVER_PYTHON"] = sys.executable

from pyspark.sql import SparkSession

spark = SparkSession.builder \
    .appName("PySparkPractice") \
    .master("local[*]") \
    .getOrCreate()

print(spark.version)
```

```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

# Aggregation

```python

emp_data = [
(1,'manish',26,20000,'india','IT'),
(2,'rahul',None,40000,'germany','engineering'),
(3,'pawan',12,60000,'india','sales'),
(4,'roshini',44,None,'uk','engineering'),
(5,'raushan',35,70000,'india','sales'),
(6,None,29,200000,'uk','IT'),
(7,'adam',37,65000,'us','IT'),
(8,'chris',16,40000,'us','sales'),
(None,None,None,None,None,None),
(7,'adam',37,65000,'us','IT')
]

schema = ['id', 'name', 'age', 'salary', 'dept']

emp_1 = spark.createDataFrame(emp_data, schema)
```

```python
emp_1.show()
```

```python
emp_1.count()
```

```python
emp_1.select('name').count()
```

```python
emp_1.select(count('name')).show()
```

```python
emp_1.select(count('*')).show()
```

```python
emp_1.select(sum('salary'), max('salary'), min('salary')).show()
```

## alias used for changing the name

```python
emp_1.select(sum('salary').alias('total_salary'), max('salary').alias('max_salary'), min('salary').alias('min_salary')).show()
```

```python
emp_1.select(sum('salary'), count('salary'), avg('salary')).show()
```

## cast used for changing decimal into int

```python
emp_1.select(sum('salary'), count('salary'), avg('salary').cast('int')).show()
```
